#Problem Definition & Dataset:

We are tasked with finding if you can use the output of multiple LLMs to correctly identify which one wrote which, as well as seeing how the user's input can help us make the predicition, possibly without looking at the output at all.

Gathered dataset from the Large Model Systems Organization's Chatbot Arena on Kaggle: https://www.kaggle.com/competitions/lmsys-chatbot-arena/

This dataset shows the result from their own arena they ran, in which they give a single prompt to two LLMs and have a human judge which of the LLMs did a better job. While this does not completely align with our problem, this is a somewhat recent (2 years) dataset that is very large (around 57000 rows). To make this better match our goal, we split every row into two, and removed the information relating to the judges, this way we just have the model used, the prompt given, and the response returned. This doubled the amount of data we have to around 110000 responses, then we decided to only look at 3 specific models, it would be unneccesarily complex to classify the response into 64 different models, we took 3 of the most common models: gpt-4-0613, claude-2.1 & llama-2-70b-chat with 6165, 5583 & 3428 responses respectfully. This provides us with the processed data we need to run the models.
